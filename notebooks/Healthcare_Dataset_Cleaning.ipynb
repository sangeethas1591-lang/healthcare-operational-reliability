Loading a dataset - Healthcare dataset

```python
from google.colab import drive
drive.mount('/content/drive')
import numpy as np
import pandas as pd
file_path = '/content/drive/MyDrive/Healthcare_Analysis_Dataset.csv'
df = pd.read_csv(file_path)

```

Inspecting the dataset loaded

```python
df.info()
```

Look up the first few rows of the dataset

```python
df.head()
```

Identification of null values in the dataset

```python
df.isnull().sum()
```

Handling Missing values (Imputation)
Check if Age is null, impute with Median value

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
df.head()
```

Datatype conversion - convert admission date to Datetime

```python
df['Date of Admission'] = pd.to_datetime(df['Date of Admission'])
df.info()
```

Creation of a Derived column - Billing Category. Categorizing billing into 'High' or 'Low' for easier visualization

```python
df['Billing_Category']= np.where(df['Billing Amount']>5000,'High','Low')
df.head()
```

Export the cleaned dataset to csv

```python
df.to_csv('Cleaned_Healthcare.csv',index=False)
print('Exported Successfully')
```
